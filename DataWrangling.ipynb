<h3>1.4 Wrangle Data for Dashboard</h3>
<h4>1.4.1 Standardize 'EdLevel'</h4>

```python
#Make EdLevel to Samll Case
df_sub['EdLevel'] = df_sub['EdLevel'].str.lower().str.strip()

#Replace instances
df_sub['EdLevel'] = df_sub['EdLevel'].replace('primary/elementary school', 'No Degree')
df_sub['EdLevel'] = df_sub['EdLevel'].replace("bachelor’s degree (b.a., b.s., b.eng., etc.)", "Bachelor's Degree")
df_sub['EdLevel'] = df_sub['EdLevel'].replace("master’s degree (m.a., m.s., m.eng., mba, etc.)", "Graduate Degree")
df_sub['EdLevel'] = df_sub['EdLevel'].replace("some college/university study without earning a degree", 'High School Degree')
df_sub['EdLevel'] = df_sub['EdLevel'].replace("secondary school (e.g. american high school, german realschule or gymnasium, etc.)", 'High School Degree')
df_sub['EdLevel'] = df_sub['EdLevel'].replace("professional degree (jd, md, ph.d, ed.d, etc.)", 'Graduate Degree')
df_sub['EdLevel'] = df_sub['EdLevel'].replace("associate degree (a.a., a.s., etc.)", 'Associate Degree')
df_sub['EdLevel'] = df_sub['EdLevel'].replace("something else", 'NA')
df_sub['EdLevel'] = df_sub['EdLevel'].fillna('NA')
```

<h4>1.4.2 Clean 'Country' Column</h4>
<p>Use Webscraping methods to crawl valid country list. Check user input against them to find spelling errors, and input errors.</p>

```python
#Find Spelling Errors/Input Errors

#import libraries for WebScraping
from bs4 import BeautifulSoup # this module helps in web scrapping.
import requests

#Scraping Country List form Worldometers.com
url = "https://www.worldometers.info/geography/alphabetical-list-of-countries/"
response = requests.get(url)
html_content = response.text

soup = BeautifulSoup(html_content, "html.parser")
#Find all <table> tags
tables = soup.find_all("table")
table = tables[0]

country = []
for row in table.find('tbody').find_all('tr'): # Assuming a tbody
    cols = row.find_all('td')
    country.append(cols[1].get_text().strip()) #Country is 2nd Column in Table
```

```python
#Replace Some Country for Standardization
country[38] = 'Congo' # From Congo (Congo-Brazzaville) 
country[44] = 'Czech Republic' #From 'Czechia (Czech Republic)'
country[56] = 'Eswatini' #From 'Eswatini (fmr. ""Swaziland"")'
country[118] = 'Myanmar' #From 'Myanmar (formerly Burma)'
country[133] = 'Palestine' #From Palestine State

#Append Country that are not in the List, but in the survey
country.append('Hong Kong (S.A.R.)')
country.append('Taiwan')
country.append('Isle of Man')
country.append('Kosovo')
country.append("Côte d'Ivoire")

#Save Country List in CSV File for future use too
pd.Series(country).to_csv('countries.csv', index=False)
```

```python
# Check if values in 'col1' are in my_list
not_valid = ~df_sub['Country'].isin(country)
df_sub[not_valid]['Country'].unique()
```

```python
#Replace Dataset for different names refering to same country
#Standardization Purpose
df_sub['Country'] = df_sub['Country'].replace('United Kingdom of Great Britain and Northern Ireland', 'United Kingdom')
df_sub['Country'] = df_sub['Country'].replace('Republic of North Macedonia', 'North Macedonia')
df_sub['Country'] = df_sub['Country'].replace('Russian Federation', 'Russia')
df_sub['Country'] = df_sub['Country'].replace('Viet Nam', 'Vietnam')
df_sub['Country'] = df_sub['Country'].replace('Venezuela, Bolivarian Republic of...', 'Venezuela')
df_sub['Country'] = df_sub['Country'].replace('United Republic of Tanzania', 'Tanzania')
df_sub['Country'] = df_sub['Country'].replace('Iran, Islamic Republic of...', 'Iran')
df_sub['Country'] = df_sub['Country'].replace('Republic of Korea', 'South Korea')
df_sub['Country'] = df_sub['Country'].replace('Syrian Arab Republic', 'Syria')
df_sub['Country'] = df_sub['Country'].replace('Republic of Moldova', 'Moldova')
df_sub['Country'] = df_sub['Country'].replace('Congo, Republic of the...', 'Congo')
df_sub['Country'] = df_sub['Country'].replace('Brunei Darussalam', 'Brunei')
df_sub['Country'] = df_sub['Country'].replace('Cape Verde', 'Cabo Verde')
```

<h4>1.4.3 Data Wrangling</h4>

```python
axis = ['Language', 'Database', 'Platform', 'Webframe']


for ax in axis:
    #To seperate data to multiple columns, use one hot encoding method
    dummies = df_sub[f'{ax}HaveWorkedWith'].str.get_dummies(sep=';')

    #Get Respondent Count for Each Category and save it to Dataframe
    lnTrend = dummies.sum().transpose().to_frame()
    
    dummies = df_sub[f'{ax}WantToWorkWith'].str.get_dummies(sep=';')
    lnTrend['InterestedUsers'] = dummies.sum().transpose().to_frame()
    lnTrend = lnTrend.reset_index()
    lnTrend.columns = ['Name', 'ExperienceUsers', 'InterestedUsers']

    #Save into .csv for usage in Dashboarding s/w
    lnTrend.to_csv(f'{ax}Trend.csv', index=False)
```
